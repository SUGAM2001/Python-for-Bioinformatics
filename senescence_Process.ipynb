```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Step 1: Define your three dataframes
proliferative_data = {
    'chr': ['chr12', 'chr22', 'chr17', 'chr6', 'chr7'],
    'locus': ['3499600-3505628', '28662883-28669715', '66596179-66602995', '2417374-2424203', '111962793-111969623'],
    'Rep1': [23.021972, 38.340212, np.nan, np.nan, np.nan],
    'Rep2': [np.nan, np.nan, 38.430804, 28.763474, 39.665118],
}

gy10_data = {
    'chr': ['chr1', 'chr2', 'chr2', 'chr3', 'chr3', 'chr6', 'chr9', 'chr12', 'chr3', 'chr7'],
    'locus': ['180866411-180873243', '148188345-148195173', '230336669-230342913', '11488272-11492890',
              '136478656-136485503', '2417374-2424203', '20655232-20659202', '3499600-3505628',
              '11488272-11492890', '111962793-111969623'],
    'Rep1': [21.349422, 29.378018, 45.487084, 25.18142, 28.441505, 20.194122, 21.118427, np.nan, np.nan, np.nan],
    'Rep2': [np.nan, np.nan, np.nan, np.nan, np.nan, np.nan, np.nan, 22.817682, 20.710568, 35.346769],
}

replicative_data = {
    'chr': ['chr2', 'chr6', 'chr12', 'chr19', 'chr3', 'chr5'],
    'locus': ['230336669-230342913', '83333552-83340381', '3499600-3505628', '43864094-43867700',
              '11488272-11492890', '34147445-34154431'],
    'Rep1': [56.741669, 65.745606, 26.983501, np.nan, np.nan, np.nan],
    'Rep2': [np.nan, np.nan, np.nan, 30.842568, 22.875997, 28.913247],
}

# Convert to DataFrames
df_prolif = pd.DataFrame(proliferative_data)
df_10gy = pd.DataFrame(gy10_data)
df_repl = pd.DataFrame(replicative_data)

# Step 2: Add condition column
df_prolif['condition'] = 'Proliferative'
df_10gy['condition'] = '10gy'
df_repl['condition'] = 'Replicative'

# Combine all into one DataFrame
df_all = pd.concat([df_prolif, df_10gy, df_repl], ignore_index=True)

# Step 3: Create full region label
df_all['region'] = df_all['chr'] + ':' + df_all['locus']

# Step 4: Compute average of Rep1 and Rep2
df_all['mean_expr'] = df_all[['Rep1', 'Rep2']].mean(axis=1)

# Step 5: Pivot for heatmap
heatmap_data = df_all.pivot_table(index='region', columns='condition', values='mean_expr')

# Optional: sort rows by mean
heatmap_data = heatmap_data.sort_index()

# Step 6: Plot
plt.figure(figsize=(12, 8))
sns.heatmap(heatmap_data, annot=True, fmt=".2f", cmap='viridis', cbar_kws={'label': 'Expression'})
#sns.heatmap(heatmap_data, annot=True, fmt=".2f", cmap='Greys', cbar_kws={'label': 'Expression'})

plt.title("Heatmap of Expression Levels Across Conditions")
plt.ylabel("Genomic Region")
plt.xlabel("Condition")
plt.tight_layout()
plt.show()
```
